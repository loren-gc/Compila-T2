````python
class Enum(tuple):
  def __getattr__(self, name):
    if name in self:
      return name
    raise AttributeError

TokenType = Enum([
  'PCDeclaracoes',
  'PCAlgoritmo',
  'PCInteiro',
  'PCReal',
  'PCAtribuir',
  'PCA',
  'PCLer',
  'PCImprimir',
  'PCSe',
  'PCEntao',
  'PCSenao',
  'PCEnquanto',
  'PCInicio',
  'PCFim',
  'OpAritMult',
  'OpAritDiv',
  'OpAritSoma',
  'OpAritSub',
  'OpRelMenor',
  'OpRelMenorIgual',
  'OpRelMaiorIgual',
  'OpRelMaior',
  'OpRelIgual',
  'OpRelDif',
  'OpLogE',
  'OpLogOu',
  'Delim',
  'AbrePar',
  'FechaPar',
  'Var',
  'NumInt',
  'NumReal',
  'Cadeia',
  'Fim'
])
````

````python
class Token:

  __slots__ = ("type", "value")

  def __init__(self, type, value):
    self.type = type
    self.value = value

  def __repr__(self):
    return '<%s,%s>' % (self.type, self.value)
````

````python
class FileStream:

  def __init__(self, path, bufsize=20):
    try:
      self.file = open(path, 'r')
      self.is_open = True
    except Exception as e:
      print(e, file=sys.stderr)
      self.is_open = False
    self._initbuf(bufsize)

  def __enter__(self):
    return self

  def __exit__(self, *args):
    if self.is_open:
      self.file.close()

  def _initbuf(self, bufsize):
    self.bufsize = bufsize
    self.buf = [ "", "" ]
    self.pos = 0
    # Como agora tem retrocesso, é necessário armazenar o
    # buffer atual, pois caso contrário ao retroceder ele
    # pode recarregar um mesmo buffer 2 vezes
    self.idx = 1
    self._loadbuf(0)

  def _loadbuf(self, idx):
    if self.is_open and self.idx != idx:
      try:
        self.idx = 1 - self.idx
        self.buf[self.idx] = self.file.read(self.bufsize)
      except Exception as e:
        print(e, file=sys.stderr)

  def _readbuf(self):
    if self.pos < self.bufsize:
      idx, pos = 0, self.pos
    else:
      idx, pos = 1, self.pos - self.bufsize
    try:
      ret = self.buf[idx][pos]
    except IndexError:
      ret = ''
    pos += 1
    if pos == self.bufsize:
      idx, pos = 1 - idx, 0
      self._loadbuf(idx)
    self.pos = idx * self.bufsize + pos
    return ret

  def read(self):
    ret = self._readbuf()
    #print(ret, end='')
    return ret

  def unread(self):
    self.pos -= 1
    if self.pos < 0:
      self.pos = self.bufsize * 2 - 1
````

````python
class LexError(Exception):
  pass
````

````python
class Lexer:

  def __init__(self, stream, skip_whitespace=True):
    self.stream = stream
    self.skip_whitespace = skip_whitespace
    self.line = 1
    self.column = 1
    self.value = ""
    self.comments = None
    self.rules = []

  def token(self):
    while True:
      self._newline()
      self._accept()
      if self.skip_whitespace:
        self._whitespace()
        self._accept()
      if self.comments is not None:
        if self.comments():
          self._accept()
          continue
      for rule in self.rules:
        ret = rule()
        if ret is None:
          self._reject()
        else:
          self._accept()
          return ret
      if self._read() == '': return Token('EOF', "")
      ret = self.error(Token('ERROR', self.value))
      if ret is not None:
        return ret

  def error(self, t):
    raise LexError("Illegal character %s at position (%d,%d)" % \
                   (repr(t.value[0]), self.line, self.column))

  def _reject(self):
    for i in range(len(self.value)):
      self.stream.unread()
    self.value = ""

  def _accept(self):
    # print(self.value, end='') # comentar para ficar melhor a saída
    if len(self.value) > 0 and self.value[-1] != '\n':
      self.column += len(self.value)
    self.value = ""

  def _read(self):
    ret = self.stream.read()
    self.value = self.value + (ret if ret != '' else ' ')
    return ret

  def _unread(self):
    self.value = self.value[:len(self.value)-1]
    self.stream.unread()

  def _newline(self):
    state = 1
    while True:
      c = self._read()
      if state == 1:
        if c == '\n':
          state = 2
          self.line += 1
          self.column = 1
        else:
          self._unread()
          return
      elif state == 2:
        if c == '\n':
          self.line += 1
          self.column = 1
        else:
          self._unread()
          return

  def _whitespace(self):
    state = 1
    while True:
      c = self._read()
      if state == 1:
        if c.isspace():
          state = 2
        else:
          self._unread()
          return
      elif state == 2:
        if not c.isspace():
          self._unread()
          return
````

````python
class AlgumaLexer(Lexer):

  def __init__(self, stream):
    super().__init__(stream)
    self.comments = self._Comentarios
    self.rules = [
      self._PC,
      self._Var,
      self._Num,
      self._Cadeia,
      self._OpArit,
      self._OpRel,
      self._Delim,
      self._Par,
      self._Fim
    ]

  def error(self, t):
    print("Illegal character %s at position (%d,%d)" % \
          (repr(t.value[0]), self.line, self.column),
         file=sys.stderr)
    self._accept()

  def _Delim(self):
    c = self._read()
    if c == ':':
      return Token(TokenType.Delim, self.value)
    else:
      return None

  def _Par(self):
    c = self._read()
    if c == '(':
      return Token(TokenType.AbrePar, self.value)
    elif c == ')':
      return Token(TokenType.FechaPar, self.value)
    else:
      return None

  def _PC(self):
    while True:
      c = self._read()
      if not c.isalpha():
        self._unread()
        if self.value == "DECLARACOES":
          return Token(TokenType.PCDeclaracoes, self.value)
        elif self.value == "ALGORITMO":
          return Token(TokenType.PCAlgoritmo, self.value)
        elif self.value == "INTEIRO":
          return Token(TokenType.PCInteiro, self.value)
        elif self.value == "REAL":
          return Token(TokenType.PCReal, self.value)
        elif self.value == "ATRIBUIR":
          return Token(TokenType.PCAtribuir, self.value)
        elif self.value == "A":
          return Token(TokenType.PCA, self.value)
        elif self.value == "LER":
          return Token(TokenType.PCLer, self.value)
        elif self.value == "IMPRIMIR":
          return Token(TokenType.PCImprimir, self.value)
        elif self.value == "SE":
          return Token(TokenType.PCSe, self.value)
        elif self.value == "ENTAO":
          return Token(TokenType.PCEntao, self.value)
        elif self.value == "SENAO":
          return Token(TokenType.PCSenao, self.value)
        elif self.value == "ENQUANTO":
          return Token(TokenType.PCEnquanto, self.value)
        elif self.value == "INICIO":
          return Token(TokenType.PCInicio, self.value)
        elif self.value == "FIM":
          return Token(TokenType.PCFim, self.value)
        elif self.value == "E":
          return Token(TokenType.OpLogE, self.value)
        elif self.value == "OU":
          return Token(TokenType.OpLogOu, self.value)
        else:
          return None

  def _OpArit(self):
    c = self._read()
    if c == '*':
      return Token(TokenType.OpAritMult, self.value)
    elif c == '/':
      return Token(TokenType.OpAritDiv, self.value)
    elif c == '+':
      return Token(TokenType.OpAritSoma, self.value)
    elif c == '-':
      return Token(TokenType.OpAritSub, self.value)
    else:
      return None

  def _OpRel(self):
    c = self._read()
    if c == '<':
      c = self._read()
      if c == '>':
        return Token(TokenType.OpRelDif, self.value)
      elif c == '=':
        return Token(TokenType.OpRelMenorIgual, self.value)
      else:
        self._unread()
        return Token(TokenType.OpRelMenor, self.value)
    elif c == '=':
      return Token(TokenType.OpRelIgual, self.value)
    elif c == '>':
      c = self._read()
      if c == '=':
        return Token(TokenType.OpRelMaiorIgual, self.value)
      else:
        self._unread()
        return Token(TokenType.OpRelMaior, self.value)
    else:
      return None

  def _Num(self):
    state = 1
    while True:
      c = self._read()
      if state == 1:
        if c.isdigit():
          state = 2
        else:
          return None
      elif state == 2:
        if c == '.':
          c = self._read()
          if c.isdigit():
            state = 3
          else:
            return None
        elif not c.isdigit():
          self._unread()
          return Token(TokenType.NumInt, self.value)
      elif state == 3:
        if not c.isdigit():
          self._unread()
          return Token(TokenType.NumReal, self.value)

  def _Var(self):
    state = 1
    while True:
      c = self._read()
      if state == 1:
        if c.isalpha():
          state = 2
        else:
          return None
      elif state == 2:
        if not c.isalpha() and not c.isdigit():
          self._unread()
          return Token(TokenType.Var, self.value)

  def _Cadeia(self):
    state = 1
    while True:
      c = self._read()
      if state == 1:
        if c == '\'':
          state = 2
        else:
          return None
      elif state == 2:
        if c == '\n':
          return None
        elif c == '\'':
          return Token(TokenType.Cadeia, self.value)
        elif c == '\\':
          state = 3
      elif state == 3:
        if c == '\n':
          return None
        else:
          state = 2

  def _Comentarios(self):
    state = 1
    while True:
      c = self._read()
      if state == 1:
        if c == '%':
          state = 2
        else:
          self._unread()
          return False
      elif state == 2:
        if c == '\n':
          self.line += 1
          self.column = 1
          return True

  def _Fim(self):
    c = self._read()
    if c == '':
      return Token(TokenType.Fim, "Fim")
    return None
````

````python
class Parser:

  def __init__(self, lexer, maxlen=10):
    self.lexer = lexer
    self.start = None
    self.stop = 'EOF'
    self.has_token = True
    self.maxlen = maxlen
    self.tokens = []
    self._read()

  def parse(self):
    if self.start is not None:
      self.start()

  def error(self, type):
    token = self._lookahead()
    if token:
      line = getattr(token, 'line', 0)
      if line:
        raise SyntaxError("Syntax error at line %d, token=%s" % \
                          (line, token.type))
      else:
        raise SyntaxError("Syntax error, token=%s" % \
                          (token.type))
    else:
      raise SyntaxError("Parse error in input. EOF")

  def _read(self):
    while self.has_token and len(self.tokens) < self.maxlen:
      t = self.lexer.token()
      self.tokens += [t]
      self.has_token = t.type != self.stop
    print("Read: %s" % (self._lookahead()))

  def _match(self, type=None):
    if self._lookahead().type == type:
      print("Match: %s" % (self._lookahead()))
      value = self._lookahead().value
      self.tokens = self.tokens[1:]
      self._read()
      return value
    else:
      self.error(type)
      return None

  def _lookahead(self, index=1):
    if len(self.tokens) == 0:
      return None
    if index > len(self.tokens):
      return self.tokens[-1]
    return self.tokens[index-1]
````

````python
class AlgumaParser(Parser):

  def __init__(self, lexer):
    super().__init__(lexer)
    self.start = self._programa
    self.stop = TokenType.Fim

  def error(self, token):
    raise SyntaxError("Expecting one of the following (%s), but it was found %s" % \
                      (", ".join(map(str, token)) if isinstance(token, list) else token, \
                       self._lookahead().type))

  # <programa> ::= ':' 'DECLARACOES' <listaDeclaracoes> ':' 'ALGORITMO' <listaComandos>
  def _programa(self):
    self._match(TokenType.Delim)
    self._match(TokenType.PCDeclaracoes)
    self._listaDeclaracoes()
    self._match(TokenType.Delim)
    self._match(TokenType.PCAlgoritmo)
    self._listaComandos()

  # <listaDeclaracoes> ::= <declaracao> <listaDeclaracoes> | <declaracao>
  def _listaDeclaracoes(self):
    # Usaremos lookahead(4)
    # Mas daria para fatorar à esquerda
    # Veja um exemplo em self._listaComandos
    if self._lookahead(4).type == TokenType.Delim:
      self._declaracao()
    elif self._lookahead(4).type == TokenType.Var:
      self._declaracao()
      self._listaDeclaracoes()
    else:
      self.error([TokenType.Delim, TokenType.Var])

  # <declaracao> ::= VARIAVEL ':' <tipoVar>
  def _declaracao(self):
    self._match(TokenType.Var)
    self._match(TokenType.Delim)
    self._tipoVar()

  # <tipoVar> ::= 'INTEIRO' | 'REAL'
  def _tipoVar(self):
    if self._lookahead().type == TokenType.PCInteiro:
      self._match(TokenType.PCInteiro)
    elif self._lookahead().type == TokenType.PCReal:
      self._match(TokenType.PCReal)

  # <expressaoAritmetica> ::= <expressaoAritmetica> '+' <termoAritmetico> | <expressaoAritmetica> '-' <termoAritmetico> | <termoAritmetico>
  # Fatorar à esquerda:
  # <expressaoAritmetica> ::= <expressaoAritmetica> ('+' <termoAritmetico> | '-' <termoAritmetico>) | <termoAritmetico>
  # Fatorar não é suficiente, pois causa loop infinito.
  # Remover a recursão à esquerda:
  # <expressaoAritmetica> ::= <termoAritmetico> <expressaoAritmetica2>
  # <expressaoAritmetica2> ::= ('+' <termoAritmetico> | '-' <termoAritmetico>) <expressaoAritmetica2> | <<empty>>
  def _expressaoAritmetica(self):
    self._termoAritmetico()
    self._expressaoAritmetica2()

  def _expressaoAritmetica2(self):
    if self._lookahead().type in [TokenType.OpAritSoma, TokenType.OpAritSub]:
      self._expressaoAritmetica2SubRegra1()
      self._expressaoAritmetica2()
    else:
      pass # empty

  def _expressaoAritmetica2SubRegra1(self):
    if self._lookahead().type == TokenType.OpAritSoma:
      self._match(TokenType.OpAritSoma)
      self._termoAritmetico()
    elif self._lookahead().type == TokenType.OpAritSub:
      self._match(TokenType.OpAritSub)
      self._termoAritmetico()
    else:
      self.error([TokenType.OpAritSoma, TokenType.OpAritSub])

  # <termoAritmetico> ::= <termoAritmetico> '*' <fatorAritmetico> | <termoAritmetico> '/' <fatorAritmetico> | <fatorAritmetico>
  # Também precisa fatorar à esquerda e eliminar recursão à esquerda:
  # <termoAritmetico> : <fatorAritmetico> <termoAritmetico2>
  # <termoAritmetico2> : ('*' <fatorAritmetico> | '/' <fatorAritmetico>) <termoAritmetico2> | <<empty>>
  def _termoAritmetico(self):
    self._fatorAritmetico()
    self._termoAritmetico2()

  def _termoAritmetico2(self):
    if self._lookahead().type in [TokenType.OpAritMult, TokenType.OpAritDiv]:
      self._termoAritmetico2SubRegra1()
      self._termoAritmetico2()
    else:
      pass # empty

  def _termoAritmetico2SubRegra1(self):
    if self._lookahead().type == TokenType.OpAritMult:
      self._match(TokenType.OpAritMult)
      self._fatorAritmetico()
    elif self._lookahead().type == TokenType.OpAritDiv:
      self._match(TokenType.OpAritDiv)
      self._fatorAritmetico()
    else:
      self.error([TokenType.OpAritMult, TokenType.OpAritDiv])

  # <fatorAritmetico> ::= NUMINT | NUMREAL | VARIAVEL | '(' <expressaoAritmetica> ')'
  def _fatorAritmetico(self):
    if self._lookahead().type == TokenType.NumInt:
      self._match(TokenType.NumInt)
    elif self._lookahead().type == TokenType.NumReal:
      self._match(TokenType.NumReal)
    elif self._lookahead().type == TokenType.Var:
      self._match(TokenType.Var)
    elif self._lookahead().type == TokenType.AbrePar:
      self._match(TokenType.AbrePar)
      self._expressaoAritmetica()
      self._match(TokenType.FechaPar)
    else:
      self.error([TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar])

  # <expressaoRelacional> ::= <expressaoRelacional> <operadorLogico> <termoRelacional> | <termoRelacional>
  # Precisa eliminar a recursão à esquerda:
  # <expressaoRelacional> ::= <termoRelacional> <expressaoRelacional2>
  # <expressaoRelacional2> ::= <operadorLogico> <termoRelacional> <expressaoRelacional2> | <<empty>>
  def _expressaoRelacional(self):
    self._termoRelacional()
    self._expressaoRelacional2()

  def _expressaoRelacional2(self):
    if self._lookahead().type in [TokenType.OpLogE, TokenType.OpLogOu]:
      self._operadorLogico()
      self._termoRelacional()
      self._expressaoRelacional2()
    else:
      pass # empty

  # <termoRelacional> ::= <expressaoAritmetica> OP_REL <expressaoAritmetica> | '(' <expressaoRelacional> ')'
  def _termoRelacional(self):
    if self._lookahead().type in [TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar]:
      self._expressaoAritmetica()
      self._operadorRelacional()
      self._expressaoAritmetica()
    else:
      self.error([TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar])

  # <operadorRelacional> ::= '=' | '<>' | '<=' | '<' | '>=' | '>'
  def _operadorRelacional(self):
    if self._lookahead().type == TokenType.OpRelDif:
      self._match(TokenType.OpRelDif)
    elif self._lookahead().type == TokenType.OpRelIgual:
      self._match(TokenType.OpRelIgual)
    elif self._lookahead().type == TokenType.OpRelMaior:
      self._match(TokenType.OpRelMaior)
    elif self._lookahead().type == TokenType.OpRelMaiorIgual:
      self._match(TokenType.OpRelMaiorIgual)
    elif self._lookahead().type == TokenType.OpRelMenor:
      self._match(TokenType.OpRelMenor)
    elif self._lookahead().type == TokenType.OpRelMenorIgual:
      self._match(TokenType.OpRelMenorIgual)
    else:
      self.error([TokenType.OpRelDif, TokenType.OpRelIgual, TokenType.OpRelMaior, TokenType.OpRelMaiorIgual, TokenType.OpRelMenor, TokenType.OpRelMenorIgual])

  # <operadorLogico> ::= 'E' | 'OU'
  def _operadorLogico(self):
    if self._lookahead().type == TokenType.OpLogE:
      self._match(TokenType.OpLogE)
    elif self._lookahead().type == TokenType.OpLogOu:
      self._match(TokenType.OpLogOu)
    else:
      self.error([TokenType.OpLogE, TokenType.OpLogOu])

  # <listaComandos> ::= <comando> <listaComandos> | <comando>
  # Vamos fatorar à esquerda:
  # <listaComandos> ::= <comando> (<listaComandos> | <<empty>>)
  def _listaComandos(self):
    self._comando()
    self._listaComandosSubRegra1()

  def _listaComandosSubRegra1(self):
    if self._lookahead().type in [TokenType.PCAtribuir, TokenType.PCLer, TokenType.PCImprimir, TokenType.PCSe, TokenType.PCEnquanto, TokenType.PCInicio]:
      self._listaComandos()
    else:
      pass # empty

  # <comando> ::= <comandoAtribuicao> | <comandoEntrada> | <comandoSaida> | <comandoCondicao> | <comandoRepeticao> | <subAlgoritmo>
  def _comando(self):
    if self._lookahead().type == TokenType.PCAtribuir:
      self._comandoAtribuicao()
    elif self._lookahead().type == TokenType.PCLer:
      self._comandoEntrada()
    elif self._lookahead().type == TokenType.PCImprimir:
      self._comandoSaida()
    elif self._lookahead().type == TokenType.PCSe:
      self._comandoCondicao()
    elif self._lookahead().type == TokenType.PCEnquanto:
      self._comandoRepeticao()
    elif self._lookahead().type == TokenType.PCInicio:
      self._subAlgoritmo()
    else:
      self.error([TokenType.PCAtribuir, TokenType.PCLer, TokenType.PCImprimir, TokenType.PCSe, TokenType.PCEnquanto, TokenType.PCInicio])

  # <comandoAtribuicao> ::= 'ATRIBUIR' <expressaoAritmetica> 'A' VARIAVEL
  def _comandoAtribuicao(self):
    self._match(TokenType.PCAtribuir)
    self._expressaoAritmetica()
    self._match(TokenType.PCA)
    self._match(TokenType.Var)

  # <comandoEntrada> : 'LER' VARIAVEL
  def _comandoEntrada(self):
    self._match(TokenType.PCLer)
    self._match(TokenType.Var)

  # <comandoSaida> : 'IMPRIMIR'  (<expressaoAritmetica> | CADEIA)
  def _comandoSaida(self):
    self._match(TokenType.PCImprimir)
    self._comandoSaidaSubRegra1()

  def _comandoSaidaSubRegra1(self):
    if self._lookahead().type in [TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar]:
      self._expressaoAritmetica()
    elif self._lookahead().type == TokenType.Cadeia:
      self._match(TokenType.Cadeia)
    else:
      self.error([TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar, TokenType.Cadeia])

  # <comandoCondicao> ::= 'SE' <expressaoRelacional> 'ENTAO' <comando> | 'SE' <expressaoRelacional> 'ENTAO' <comando> 'SENAO' <comando>
  # Fatorar à esquerda:
  # <comandoCondicao> : 'SE' <expressaoRelacional> 'ENTAO' <comando> ('SENAO' <comando> | <<empty>>)
  def _comandoCondicao(self):
    self._match(TokenType.PCSe)
    self._expressaoRelacional()
    self._match(TokenType.PCEntao)
    self._comando()
    self._comandoCondicaoSubRegra1()

  def _comandoCondicaoSubRegra1(self):
    if self._lookahead().type == TokenType.PCSenao:
      self._match(TokenType.PCSenao)
      self._comando()
    else:
      pass # empty

  # <comandoRepeticao> ::= 'ENQUANTO' <expressaoRelacional> <comando>
  def _comandoRepeticao(self):
    self._match(TokenType.PCEnquanto)
    self._expressaoRelacional()
    self._comando()

  # <subAlgoritmo> ::= 'INICIO' <listaComandos> 'FIM'
  def _subAlgoritmo(self):
    self._match(TokenType.PCInicio)
    self._listaComandos()
    self._match(TokenType.PCFim)
````

````python
class AlgumaParser(Parser):

  def __init__(self, lexer):
    super().__init__(lexer)
    self.start = self._programa
    self.stop = TokenType.Fim

  def error(self, token):
    raise SyntaxError("Expecting one of the following (%s), but it was found %s" % \
                      (", ".join(map(str, token)) if isinstance(token, list) else token, \
                       self._lookahead().type))

  # <programa> ::= ':' 'DECLARACOES' <listaDeclaracoes> ':' 'ALGORITMO' <listaComandos> EOF
  def _programa(self):
    self._match(TokenType.Delim)
    self._match(TokenType.PCDeclaracoes)
    self._listaDeclaracoes()
    self._match(TokenType.Delim)
    self._match(TokenType.PCAlgoritmo)
    self._listaComandos()
    self._match(TokenType.Fim)

  # <listaDeclaracoes> ::= <declaracao> <listaDeclaracoes> | <declaracao>
  def _listaDeclaracoes(self):
    # Usaremos lookahead(4)
    # Mas daria para fatorar à esquerda
    # Veja um exemplo em self._listaComandos
    if self._lookahead(4).type == TokenType.Delim:
      self._declaracao()
    elif self._lookahead(4).type == TokenType.Var:
      self._declaracao()
      self._listaDeclaracoes()
    else:
      self.error([TokenType.Delim, TokenType.Var])

  # <declaracao> ::= VARIAVEL ':' <tipoVar>
  def _declaracao(self):
    self._match(TokenType.Var)
    self._match(TokenType.Delim)
    self._tipoVar()

  # <tipoVar> ::= 'INTEIRO' | 'REAL'
  def _tipoVar(self):
    if self._lookahead().type == TokenType.PCInteiro:
      self._match(TokenType.PCInteiro)
    elif self._lookahead().type == TokenType.PCReal:
      self._match(TokenType.PCReal)

  # <expressaoAritmetica> ::= <expressaoAritmetica> '+' <termoAritmetico> | <expressaoAritmetica> '-' <termoAritmetico> | <termoAritmetico>
  # Fatorar à esquerda:
  # <expressaoAritmetica> ::= <expressaoAritmetica> ('+' <termoAritmetico> | '-' <termoAritmetico>) | <termoAritmetico>
  # Fatorar não é suficiente, pois causa loop infinito.
  # Remover a recursão à esquerda:
  # <expressaoAritmetica> ::= <termoAritmetico> <expressaoAritmetica2>
  # <expressaoAritmetica2> ::= ('+' <termoAritmetico> | '-' <termoAritmetico>) <expressaoAritmetica2> | <<empty>>
  def _expressaoAritmetica(self):
    self._termoAritmetico()
    self._expressaoAritmetica2()

  def _expressaoAritmetica2(self):
    if self._lookahead().type in [TokenType.OpAritSoma, TokenType.OpAritSub]:
      self._expressaoAritmetica2SubRegra1()
      self._expressaoAritmetica2()
    else:
      pass # empty

  def _expressaoAritmetica2SubRegra1(self):
    if self._lookahead().type == TokenType.OpAritSoma:
      self._match(TokenType.OpAritSoma)
      self._termoAritmetico()
    elif self._lookahead().type == TokenType.OpAritSub:
      self._match(TokenType.OpAritSub)
      self._termoAritmetico()
    else:
      self.error([TokenType.OpAritSoma, TokenType.OpAritSub])

  # <termoAritmetico> ::= <termoAritmetico> '*' <fatorAritmetico> | <termoAritmetico> '/' <fatorAritmetico> | <fatorAritmetico>
  # Também precisa fatorar à esquerda e eliminar recursão à esquerda:
  # <termoAritmetico> : <fatorAritmetico> <termoAritmetico2>
  # <termoAritmetico2> : ('*' <fatorAritmetico> | '/' <fatorAritmetico>) <termoAritmetico2> | <<empty>>
  def _termoAritmetico(self):
    self._fatorAritmetico()
    self._termoAritmetico2()

  def _termoAritmetico2(self):
    if self._lookahead().type in [TokenType.OpAritMult, TokenType.OpAritDiv]:
      self._termoAritmetico2SubRegra1()
      self._termoAritmetico2()
    else:
      pass # empty

  def _termoAritmetico2SubRegra1(self):
    if self._lookahead().type == TokenType.OpAritMult:
      self._match(TokenType.OpAritMult)
      self._fatorAritmetico()
    elif self._lookahead().type == TokenType.OpAritDiv:
      self._match(TokenType.OpAritDiv)
      self._fatorAritmetico()
    else:
      self.error([TokenType.OpAritMult, TokenType.OpAritDiv])

  # <fatorAritmetico> ::= NUMINT | NUMREAL | VARIAVEL | '(' <expressaoAritmetica> ')'
  def _fatorAritmetico(self):
    if self._lookahead().type == TokenType.NumInt:
      self._match(TokenType.NumInt)
    elif self._lookahead().type == TokenType.NumReal:
      self._match(TokenType.NumReal)
    elif self._lookahead().type == TokenType.Var:
      self._match(TokenType.Var)
    elif self._lookahead().type == TokenType.AbrePar:
      self._match(TokenType.AbrePar)
      self._expressaoAritmetica()
      self._match(TokenType.FechaPar)
    else:
      self.error([TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar])

  # <expressaoRelacional> ::= <expressaoRelacional> <operadorLogico> <termoRelacional> | <termoRelacional>
  # Precisa eliminar a recursão à esquerda:
  # <expressaoRelacional> ::= <termoRelacional> <expressaoRelacional2>
  # <expressaoRelacional2> ::= <operadorLogico> <termoRelacional> <expressaoRelacional2> | <<empty>>
  def _expressaoRelacional(self):
    self._termoRelacional()
    self._expressaoRelacional2()

  def _expressaoRelacional2(self):
    if self._lookahead().type in [TokenType.OpLogE, TokenType.OpLogOu]:
      self._operadorLogico()
      self._termoRelacional()
      self._expressaoRelacional2()
    else:
      pass # empty

  # <termoRelacional> ::= <expressaoAritmetica> OP_REL <expressaoAritmetica> | '(' <expressaoRelacional> ')'
  def _termoRelacional(self):
    if self._lookahead().type in [TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar]:
      self._expressaoAritmetica()
      self._operadorRelacional()
      self._expressaoAritmetica()
    else:
      self.error([TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar])

  # <operadorRelacional> ::= '=' | '<>' | '<=' | '<' | '>=' | '>'
  def _operadorRelacional(self):
    if self._lookahead().type == TokenType.OpRelDif:
      self._match(TokenType.OpRelDif)
    elif self._lookahead().type == TokenType.OpRelIgual:
      self._match(TokenType.OpRelIgual)
    elif self._lookahead().type == TokenType.OpRelMaior:
      self._match(TokenType.OpRelMaior)
    elif self._lookahead().type == TokenType.OpRelMaiorIgual:
      self._match(TokenType.OpRelMaiorIgual)
    elif self._lookahead().type == TokenType.OpRelMenor:
      self._match(TokenType.OpRelMenor)
    elif self._lookahead().type == TokenType.OpRelMenorIgual:
      self._match(TokenType.OpRelMenorIgual)
    else:
      self.error([TokenType.OpRelDif, TokenType.OpRelIgual, TokenType.OpRelMaior, TokenType.OpRelMaiorIgual, TokenType.OpRelMenor, TokenType.OpRelMenorIgual])

  # <operadorLogico> ::= 'E' | 'OU'
  def _operadorLogico(self):
    if self._lookahead().type == TokenType.OpLogE:
      self._match(TokenType.OpLogE)
    elif self._lookahead().type == TokenType.OpLogOu:
      self._match(TokenType.OpLogOu)
    else:
      self.error([TokenType.OpLogE, TokenType.OpLogOu])

  # <listaComandos> ::= <comando> <listaComandos> | <comando>
  # Vamos fatorar à esquerda:
  # <listaComandos> ::= <comando> (<listaComandos> | <<empty>>)
  def _listaComandos(self):
    self._comando()
    self._listaComandosSubRegra1()

  def _listaComandosSubRegra1(self):
    if self._lookahead().type in [TokenType.PCAtribuir, TokenType.PCLer, TokenType.PCImprimir, TokenType.PCSe, TokenType.PCEnquanto, TokenType.PCInicio]:
      self._listaComandos()
    else:
      pass # empty

  # <comando> ::= <comandoAtribuicao> | <comandoEntrada> | <comandoSaida> | <comandoCondicao> | <comandoRepeticao> | <subAlgoritmo>
  def _comando(self):
    if self._lookahead().type == TokenType.PCAtribuir:
      self._comandoAtribuicao()
    elif self._lookahead().type == TokenType.PCLer:
      self._comandoEntrada()
    elif self._lookahead().type == TokenType.PCImprimir:
      self._comandoSaida()
    elif self._lookahead().type == TokenType.PCSe:
      self._comandoCondicao()
    elif self._lookahead().type == TokenType.PCEnquanto:
      self._comandoRepeticao()
    elif self._lookahead().type == TokenType.PCInicio:
      self._subAlgoritmo()
    else:
      self.error([TokenType.PCAtribuir, TokenType.PCLer, TokenType.PCImprimir, TokenType.PCSe, TokenType.PCEnquanto, TokenType.PCInicio])

  # <comandoAtribuicao> ::= 'ATRIBUIR' <expressaoAritmetica> 'A' VARIAVEL
  def _comandoAtribuicao(self):
    self._match(TokenType.PCAtribuir)
    self._expressaoAritmetica()
    self._match(TokenType.PCA)
    self._match(TokenType.Var)

  # <comandoEntrada> : 'LER' VARIAVEL
  def _comandoEntrada(self):
    self._match(TokenType.PCLer)
    self._match(TokenType.Var)

  # <comandoSaida> : 'IMPRIMIR'  (<expressaoAritmetica> | CADEIA)
  def _comandoSaida(self):
    self._match(TokenType.PCImprimir)
    self._comandoSaidaSubRegra1()

  def _comandoSaidaSubRegra1(self):
    if self._lookahead().type in [TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar]:
      self._expressaoAritmetica()
    elif self._lookahead().type == TokenType.Cadeia:
      self._match(TokenType.Cadeia)
    else:
      self.error([TokenType.NumInt, TokenType.NumReal, TokenType.Var, TokenType.AbrePar, TokenType.Cadeia])

  # <comandoCondicao> ::= 'SE' <expressaoRelacional> 'ENTAO' <comando> | 'SE' <expressaoRelacional> 'ENTAO' <comando> 'SENAO' <comando>
  # Fatorar à esquerda:
  # <comandoCondicao> : 'SE' <expressaoRelacional> 'ENTAO' <comando> ('SENAO' <comando> | <<empty>>)
  def _comandoCondicao(self):
    self._match(TokenType.PCSe)
    self._expressaoRelacional()
    self._match(TokenType.PCEntao)
    self._comando()
    self._comandoCondicaoSubRegra1()

  def _comandoCondicaoSubRegra1(self):
    if self._lookahead().type == TokenType.PCSenao:
      self._match(TokenType.PCSenao)
      self._comando()
    else:
      pass # empty

  # <comandoRepeticao> ::= 'ENQUANTO' <expressaoRelacional> <comando>
  def _comandoRepeticao(self):
    self._match(TokenType.PCEnquanto)
    self._expressaoRelacional()
    self._comando()

  # <subAlgoritmo> ::= 'INICIO' <listaComandos> 'FIM'
  def _subAlgoritmo(self):
    self._match(TokenType.PCInicio)
    self._listaComandos()
    self._match(TokenType.PCFim)
````

````python
class AlgumaLexer(Lexer):
  """A lexer for the ALGUMA language."""

  # Reserved keywords
  keywords = {
    'DECLARACOES': "PCDeclaracoes",
    'ALGORITMO': "PCAlgoritmo",
    'INTEIRO': "PCInteiro",
    'REAL': "PCReal",
    'ATRIBUIR': "PCAtribuir",
    'A': "PCA",
    'LER': "PCLer",
    'IMPRIMIR': "PCImprimir",
    'SE': "PCSe",
    'ENTAO': "PCEntao",
    'SENAO': "PCSenao",
    'ENQUANTO': "PCEnquanto",
    'INICIO': "PCInicio",
    'FIM': "PCFim",
    'E': "OpLogE",
    'OU': "OpLogOu",
  }

  # All the tokens recognized by the lexer
  tokens = tuple(keywords.values()) + (
    # Identifiers
    "Var",
    # Constants
    "NumInt",
    "NumReal",
    "Cadeia",
    # Operators
    "OpAritSoma",
    "OpAritSub",
    "OpAritMult",
    "OpAritDiv",
    "OpRelIgual",
    "OpRelDif",
    "OpRelMenorIgual",
    "OpRelMenor",
    "OpRelMaiorIgual",
    "OpRelMaior",
    # Delimeters
    "Delim",
    "AbrePar",
    "FechaPar",
  )

  # String containing ignored characters (between tokens)
  ignore = ' \t'

  # Other ignored patterns
  ignore_newline = r'\n+'
  ignore_comment = r'\%.*\r?\n'

  # Regular expression rules for tokens
  Var = r'[a-zA-Z][a-zA-Z0-9]*'
  NumReal = r'\d+\.\d+'
  NumInt = r'\d+'
  Cadeia = r'\'([^\\\n]|(\\.))*?\''
  OpAritSoma = r'\+'
  OpAritSub = r'-'
  OpAritMult = r'\*'
  OpAritDiv = r'\/'
  OpRelIgual = r'='
  OpRelDif = r'<>'
  OpRelMenorIgual = r'<='
  OpRelMenor = r'<'
  OpRelMaiorIgual = r'>='
  OpRelMaior = r'>'
  Delim = r':'
  AbrePar = r'\('
  FechaPar = r'\)'

  # Special cases
  def Var(self, t):
    t.type = self.keywords.get(t.value, "Var")
    return t

  # Define a rule so we can track line numbers
  def ignore_newline(self, t):
    self.lineno += len(t.value)

  def ignore_comment(self, t):
    self.lineno += t.value.count("\n")

  def find_column(self, token):
    """Find the column of the token in its line."""
    last_cr = self.text.rfind('\n', 0, token.index)
    return token.index - last_cr

  # Error handling rule
  def error(self, t):
    print("Illegal character %s at position (%d,%d)" % \
          (repr(t.value[0]), t.lineno, self.find_column(t)))
    self.index += 1
````

````python
def build_tree(root):
  return '\n'.join(_build_tree(root))

def _build_tree(node):
  if isinstance(node, list):
    if not node: return
    node = tuple(node)

  if not isinstance(node, tuple):
      yield " "+str(node)
      return

  values = [_build_tree(n) for n in node]
  if len(values) == 1:
      yield from build_lines('──', '  ', values[0])
      return

  start, *mid, end = values
  yield from build_lines('┬─', '│ ', start)
  for value in mid:
    yield from build_lines('├─', '│ ', value)
  yield from build_lines('└─', '  ', end)

def build_lines(first, other, values):
  try:
    yield first + next(values)
    for value in values:
      yield other + value
  except StopIteration:
    return
````
